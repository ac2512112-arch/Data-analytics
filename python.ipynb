```python
category_revenue = (
    df.groupby("category", as_index=False)["total_price"]          #category revenue
      .sum()
      .rename(columns={"total_price": "total_revenue"})
)

print(category_revenue)
```

```python
sorted_df = df.sort_values(
    by=["order_date", "total_price", "category"],       #sort multiple columns
    ascending=[True, False,True]
)

print(sorted_df)
```

```python
# Select numerical columns
numeric_df = df.select_dtypes(include="number")

# Create correlation matrix                       #correlation matrix
correlation_matrix = numeric_df.corr()

print(correlation_matrix)
```
